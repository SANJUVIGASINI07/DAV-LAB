```python
import pandas as pd

# Create an inbuilt dataset
data = {
    "ID": [1, 2, 3, 4, 5, 6, 7, 8],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Helen"],
    "Age": [21, 22, 20, 23, 21, 24, 22, 23],
    "Marks": [85, 67, 92, 74, 88, 55, 79, 95],
    "Category": ["A", "B", "A", "B", "A", "C", "B", "A"]
}

# Convert dictionary to DataFrame
df = pd.DataFrame(data)

# Display first and last few rows
print("First 5 rows:\n")
print(df.head())

print("\nLast 5 rows:\n")
print(df.tail())

# Check data types and general info
print("\nDataFrame Info:")
df.info()

# Summary statistics
print("\nSummary Statistics:")
print(df.describe())

# Handle missing values
df.fillna(df.mean(numeric_only=True), inplace=True)

# Create a new column
df["Double_Marks"] = df["Marks"] * 2

# Create a Series and perform operations
series = df["Marks"]
print("\nSeries + 10:")
print(series + 10)

# Filter rows
filtered_df = df[(df["Marks"] > 50) & (df["Age"] < 23)]
print("\nFiltered DataFrame:")
print(filtered_df)

# Grouping and aggregation
grouped = df.groupby("Category")["Marks"].mean()
print("\nAverage Marks by Category:")
print(grouped)

# Sorting
df_sorted = df.sort_values(by="Marks", ascending=False)
print("\nSorted DataFrame:")
print(df_sorted)

# Boolean masking
masked_df = df[df["Marks"] > df["Marks"].median()]
print("\nMarks greater than median:")
print(masked_df)

# Remove duplicates
df.drop_duplicates(inplace=True)

# Drop missing values
df.dropna(inplace=True)

# Create subset
subset_df = df[["Name", "Marks"]]
print("\nSubset DataFrame:")
print(subset_df)

# Save to CSV
subset_df.to_csv("filtered_data.csv", index=False)
print("\nCSV file 'filtered_data.csv' created successfully.")

# Compute summary statistics
print("\nTotal Marks:", df["Marks"].sum())
print("Mean Marks:", df["Marks"].mean())
print("Standard Deviation:", df["Marks"].std())
```
