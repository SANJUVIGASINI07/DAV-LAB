```python
import pandas as pd

# Read data from a website
web_df = pd.read_csv(
    "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
)

# Create sample text DataFrame
text_df = pd.DataFrame({
    "Name": ["Alice", "Bob", None, "David", "Eva"],
    "Age": [21, 22, 23, None, 25],
    "Marks": [85, None, 90, 78, 88]
})

# Create sample Excel DataFrame
excel_df = pd.DataFrame({
    "Product": ["Pen", "Book", "Pencil", None, "Scale"],
    "Price": [10, 50, None, 20, 15],
    "Quantity": [100, None, 150, 200, 120]
})

# Display data
print("Text DataFrame:")
print(text_df.head())

print("\nExcel DataFrame:")
print(excel_df.head())

print("\nWeb DataFrame:")
print(web_df.head())

# Handle missing values
text_df.ffill(inplace=True)     # Forward Fill
excel_df.bfill(inplace=True)    # Backward Fill
web_df.dropna(inplace=True)

# Save processed data
text_df.to_csv("processed_text.csv", index=False)
excel_df.to_excel("processed_excel.xlsx", index=False)

print("\nFiles saved successfully.")
```
