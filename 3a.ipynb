```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis
from sklearn.datasets import load_diabetes

# Load built-in Diabetes dataset
diabetes = load_diabetes()

# Convert to DataFrame
df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)

# Rename columns for demonstration
df.rename(columns={
    'bmi': 'BMI',
    'age': 'Age'
}, inplace=True)

# Create additional columns (for demonstration only)
np.random.seed(42)
df["Glucose"] = np.random.randint(70, 180, len(df))
df["BloodPressure"] = np.random.randint(60, 120, len(df))
df["SkinThickness"] = np.random.randint(10, 50, len(df))
df["Insulin"] = np.random.randint(15, 300, len(df))
df["DiabetesPedigreeFunction"] = np.random.uniform(0.1, 2.5, len(df))

# Display sample data
print("Sample Dataset:")
print(df.head())

# Relevant numerical columns
numerical_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

# Univariate Analysis Function
def univariate_analysis(df, columns):
    stats = {}

    for col in columns:
        stats[col] = {
            "Mean": np.mean(df[col]),
            "Median": np.median(df[col]),
            "Mode": df[col].mode()[0],
            "Variance": np.var(df[col], ddof=1),
            "Standard Deviation": np.std(df[col], ddof=1),
            "Skewness": skew(df[col]),
            "Kurtosis": kurtosis(df[col])
        }

    return pd.DataFrame(stats).T

# Perform Univariate Analysis
result = univariate_analysis(df, numerical_columns)

# Display Results
print("\nUnivariate Analysis Statistics:")
print(result)
```
