```python
import pandas as pd
import numpy as np

# Create sample statistics for UCI Diabetes Dataset
uci_stats = pd.DataFrame({
    "Mean": [120.5, 72.4, 31.8, 45.2],
    "Median": [118.0, 72.0, 31.5, 44.0],
    "Std Dev": [30.1, 12.5, 6.8, 13.4]
}, index=["Glucose", "BloodPressure", "BMI", "Age"])

# Create sample statistics for Pima Diabetes Dataset
pima_stats = pd.DataFrame({
    "Mean": [115.8, 70.3, 30.5, 43.8],
    "Median": [114.0, 70.0, 30.0, 43.0],
    "Std Dev": [28.4, 11.8, 6.2, 12.7]
}, index=["Glucose", "BloodPressure", "BMI", "Age"])

# Display Summary Statistics
print("Comparison of Univariate Analysis Results")

print("\nUCI Diabetes Dataset Statistics:")
print(uci_stats)

print("\nPima Indians Diabetes Dataset Statistics:")
print(pima_stats)

# Example Regression Performance
uci_r2 = 0.78
pima_r2 = 0.72

uci_accuracy = 82.4
pima_accuracy = 79.1

print("\nRegression Model Comparison")
print("---------------------------------")
print(f"Linear Regression R² Score")
print(f"UCI Dataset   : {uci_r2}")
print(f"Pima Dataset  : {pima_r2}")

print("\nLogistic Regression Accuracy")
print(f"UCI Dataset   : {uci_accuracy}%")
print(f"Pima Dataset  : {pima_accuracy}%")

# Compare Performance
print("\nPerformance Analysis")
if uci_r2 > pima_r2:
    print("UCI dataset has better Linear Regression performance.")
else:
    print("Pima dataset has better Linear Regression performance.")

if uci_accuracy > pima_accuracy:
    print("UCI dataset has better Logistic Regression accuracy.")
else:
    print("Pima dataset has better Logistic Regression accuracy.")
```
