```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
from sklearn.datasets import load_diabetes

# Load built-in diabetes dataset
diabetes = load_diabetes()

# Create DataFrame
uci_diabetes = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)

# Rename BMI column
uci_diabetes.rename(columns={"bmi": "BMI"}, inplace=True)

# Create a sample Glucose column (for demonstration)
np.random.seed(42)
uci_diabetes["Glucose"] = np.random.randint(70, 180, len(uci_diabetes))

# Set figure size
plt.figure(figsize=(12, 5))

# ------------------ Normal Curve for Glucose ------------------
plt.subplot(1, 2, 1)

sns.histplot(
    uci_diabetes["Glucose"],
    kde=True,
    stat="density",
    bins=20
)

x = np.linspace(
    uci_diabetes["Glucose"].min(),
    uci_diabetes["Glucose"].max(),
    100
)

plt.plot(
    x,
    norm.pdf(
        x,
        uci_diabetes["Glucose"].mean(),
        uci_diabetes["Glucose"].std()
    ),
    color="red",
    linewidth=2
)

plt.title("Normal Curve - Glucose")
plt.xlabel("Glucose")
plt.ylabel("Density")

# ------------------ Normal Curve for BMI ------------------
plt.subplot(1, 2, 2)

sns.histplot(
    uci_diabetes["BMI"],
    kde=True,
    stat="density",
    bins=20
)

x = np.linspace(
    uci_diabetes["BMI"].min(),
    uci_diabetes["BMI"].max(),
    100
)

plt.plot(
    x,
    norm.pdf(
        x,
        uci_diabetes["BMI"].mean(),
        uci_diabetes["BMI"].std()
    ),
    color="red",
    linewidth=2
)

plt.title("Normal Curve - BMI")
plt.xlabel("BMI")
plt.ylabel("Density")

plt.tight_layout()
plt.show()
```
