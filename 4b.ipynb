```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from statsmodels.stats.weightstats import ztest

# Load built-in diabetes dataset
diabetes = load_diabetes()

# Create DataFrame
uci_diabetes = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)

# Create a sample Glucose column
np.random.seed(42)
uci_diabetes["Glucose"] = np.random.randint(70, 180, len(uci_diabetes))

# Perform Z-Test (Testing if mean Glucose differs from 100)
z_stat, p_value = ztest(uci_diabetes["Glucose"], value=100)

# Display Results
print(f"Z-Statistic: {z_stat:.4f}")
print(f"P-Value: {p_value:.4f}")

# Interpretation
alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis: The mean Glucose level is significantly different from 100.")
else:
    print("Fail to reject the null hypothesis: No significant difference in mean Glucose level.")
```
