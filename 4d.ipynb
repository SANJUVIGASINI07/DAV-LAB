```python
import pandas as pd
import numpy as np
from scipy.stats import f_oneway
from sklearn.datasets import load_diabetes

# Load built-in Diabetes dataset
diabetes = load_diabetes()

# Convert to DataFrame
df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)

# Rename columns
df.rename(columns={"bmi": "BMI", "age": "Age"}, inplace=True)

# Create sample Glucose and BloodPressure columns
np.random.seed(42)
df["Glucose"] = np.random.randint(70, 180, len(df))
df["BloodPressure"] = np.random.randint(60, 120, len(df))

# Split into two datasets (simulating UCI and Pima datasets)
uci_diabetes = df.iloc[:221].copy()
pima_diabetes = df.iloc[221:].copy()

# Display sample data
print("UCI Diabetes Dataset Sample:")
print(uci_diabetes.head())

print("\nPima Diabetes Dataset Sample:")
print(pima_diabetes.head())

# Select relevant numerical columns
numerical_columns = ["Glucose", "BloodPressure", "BMI"]

# Perform One-Way ANOVA
anova_results = {}

for col in numerical_columns:
    f_stat, p_value = f_oneway(
        uci_diabetes[col],
        pima_diabetes[col]
    )

    anova_results[col] = {
        "F-statistic": f_stat,
        "P-value": p_value
    }

# Convert results to DataFrame
anova_df = pd.DataFrame(anova_results).T

# Display results
print("\nANOVA Results:")
print(anova_df)

# Interpretation
print("\nInterpretation:")
alpha = 0.05

for col in numerical_columns:
    if anova_results[col]["P-value"] < alpha:
        print(f"{col}: Significant difference between the two datasets.")
    else:
        print(f"{col}: No significant difference between the two datasets.")
```
